```python
def depositar(saldo):
  deposito = float(input("Digite o valor do depósito: "))
  if deposito > 0:
    saldo = saldo + deposito
    print(f"Depósito de R$ {deposito:.2f} realizado com sucesso.")
    print(f"\nSeu saldo atual é: R$ {saldo:.2f}")
  else:
    print("Valor de depósito inválido.")
  return saldo

def sacar(saldo):
  valor_saque = float(input("Digite o valor do saque: "))
  if valor_saque > saldo:
    print("Saldo insuficiente para realizar o saque.")
  elif valor_saque <= 0:
    print("Valor de saque inválido.")
  else:
    saldo = saldo - valor_saque
    print(f"Saque de R$ {valor_saque:.2f} realizado com sucesso.")
    print(f"\nSeu saldo atual é: R$ {saldo:.2f}")
  return saldo
```

```python
print("=== CAIXA ELETRÔNICO GENERATION ===")
saldo = float(input("Digite seu saldo inicial: "))

while True:
    print("\n=== CAIXA ELETRÔNICO GENERETION ===")
    print("1 - Checar saldo")
    print("2 - Depositar")
    print("3 - Sacar")
    print("4 - Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        print(f"\nSeu saldo atual é: R$ {saldo}")
    elif opcao == "2":
        saldo = depositar(saldo)
    elif opcao == "3":
        saldo = sacar(saldo)
    elif opcao == "4":
        print("")
        break
    else:
        print("CALMA AMIGAO E SO UM NUMERO")
```
