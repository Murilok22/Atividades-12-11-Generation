```python
def celcius_f(temperatura):
  # a conta para conversao e f = (valor em °C * 1.8 + 32)
  f = (temperatura * 1.8) + 32
  return (f)

def celcius_k(temperatura):
  # a conta para conversao e k = valor em °C + 273
  K = temperatura + 273
  return (K)
```

```python
temperatura = float(input("Digite a temperatura em °C: "))
print(celcius_f(temperatura))
print(celcius_k(temperatura))
```
